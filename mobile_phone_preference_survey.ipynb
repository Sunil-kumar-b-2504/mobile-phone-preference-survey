```python
import pandas as pd
import matplotlib.pyplot as plt

# Load your CSV file
df = pd.read_csv("survey.csv")

# Show first few rows
print(df.head(5))


```

```python
print("\nDataset Information:")
print(df.info())
```

```python
# Average Screen Time
plt.figure(figsize=(6,4))
df['8. what is your average daily screen time ?'].value_counts().plot(kind='bar')
plt.title("Average Daily Screen Time")
plt.xlabel("Hours")
plt.ylabel("Number of Users")
plt.show()
```

```python
# 1️⃣ Most Preferred Mobile Brand
plt.figure(figsize=(6,4))
df['4. the name of the mobile phone brand your are using right now'].value_counts().plot(kind='bar')
plt.title("Most Preferred Mobile Brands")
plt.xlabel("Brand")
plt.ylabel("Number of Users")
plt.show()

```

```python
# 2️⃣ Preferred Operating System
plt.figure(figsize=(5,5))
df['7. what is your preferred os ?'].value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.title("Android vs iOS Preference")
plt.ylabel("")
plt.show()

```

```python
# Cost of Current Mobile
df['9. what is the cost of your current mobile ?'].value_counts().plot(kind='bar', color='green')
plt.title("Cost of Current Mobile")
plt.xlabel("Price Range")
plt.ylabel("Number of Users")
plt.show()
```

```python

feature_counts = df['10. what feature you mainly look up to while buying a phone?'].value_counts()

# Create pie chart
plt.figure(figsize=(6, 6))
plt.pie(feature_counts, labels=feature_counts.index, autopct='%1.1f%%', startangle=140)
plt.title("Main Feature People Look for While Buying a Phone")
plt.show()
```

```python
#  Planning to Upgrade Next Year
df['15. are you planning to upgrade your phone next year?'].value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.title("Planning to Upgrade Phone Next Year")
plt.ylabel("")
plt.show()
```

Around 50% of users own phones costing between ₹20,000–₹40,000.

A small percentage have phones above ₹40,000.

60–70% of respondents plan to upgrade their phones next year, mainly for better camera and performance.

Battery life and camera quality were the most valued features among all respondents.

Android users dominate the sample, while iOS users form a smaller group.
